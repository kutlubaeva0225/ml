```python
import numpy as np
import matplotlib.pyplot as plt
import plotly.graph_objects as go

prc = 100
a = -1
b = 1
x0 = 0
sig = 0.5
x1 = 1
sig1 = 1

A1 = np.random.uniform(low=a, high=b, size=prc)
B1 = np.random.uniform(low=a, high=b, size=prc)
E1 = np.random.uniform(low=a, high=b, size=prc)

tb1 = np.column_stack((A1 - 0.5, B1, 0.5 + A1 + B1 + E1)) #матрица 100 на 3

A2 = np.random.normal(x0, sig, prc)
B2 = np.random.normal(x0, sig, prc)
E2 = np.random.normal(x0, sig, prc)

tb2 = np.column_stack((A2, 0.1*B2, 0.1*A2 + 0.2* B2 + 0.1*E2)) #матрица 100 на 3

X = np.random.normal(x1, sig1, prc)
Y = np.random.uniform(a, b, prc) 

tb3 = np.column_stack((X, Y, X**2 + Y)) #матрица 100 на 3



fig = go.Figure()

fig.add_trace(go.Scatter3d(
    x=A1 - 0.5,
    y=B1,
    z=0.5 + A1 + B1 + E1,
    mode='markers',
    marker=dict(size=4, color='royalblue', opacity=0.8),
    name='prec 1'
))
fig.add_trace(go.Scatter3d(
    x=A2,
    y=0.1 * B2,
    z=0.1 * A2 + 0.2 * B2 + 0.1 * E2,
    mode='markers',
    marker=dict(size=4, color='darkorange', opacity=0.8),
    name='prec 2'
))
fig.add_trace(go.Scatter3d(
    x=X,
    y=Y,
    z=X**2 + Y,
    mode='markers',
    marker=dict(size=4, color='forestgreen', opacity=0.8),
    name='prec 3'
))
fig.update_layout(
    title='3D график трех выборок',
    scene=dict(
        xaxis_title='x',
        yaxis_title='y',
        zaxis_title='z'
    ),
    width=900,
    height=700,
    margin=dict(l=0, r=0, b=0, t=40)
)

fig.show()
```

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

lab1 = np.zeros(len(tb1))
lab2 = np.ones(len(tb2))
lab3 = np.ones(len(tb3))
Xt = np.vstack([tb1, tb2, tb3])
Yt = np.concatenate([lab1, lab2, lab3])

x_tr, xt, y_tr, yt = train_test_split(Xt, Yt, test_size=0.2, random_state=9) #парметр random_state определяет способ выбора 20 и 80 процентов из выборки. Генератор случайных числев будет стартовать с всегда с 9

scaler = StandardScaler()
x_tr_sc = scaler.fit_transform(x_tr)
xt_sc = scaler.fit_transform(xt)
```

```python
from sklearn.svm import SVC
from sklearn.mixture import GaussianMixture
from sklearn.gaussian_process.kernels import RBF
from sklearn.gaussian_process import GaussianProcessClassifier

md1 = SVC(kernel='linear', C=0.7)
md1.fit(x_tr_sc, y_tr)

gss_0 = GaussianMixture(n_components= 1, covariance_type='full', random_state=9) # Внутри GMM используется итеративный алгоритм (EM-алгоритм), который начинает поиск с выбора случайных начальных положений центров. Число 9 фиксирует этот генератор случайных чисел.
gss_0.fit(x_tr_sc[y_tr == 0])

gss_1 = GaussianMixture(n_components=2, covariance_type='full', random_state=9)
gss_1.fit(x_tr_sc[y_tr == 1])

ker = 1.0 * RBF(length_scale=1.0)

gpc = GaussianProcessClassifier(kernel=ker, random_state=9)
gpc.fit(x_tr_sc, y_tr)
```

```python
from sklearn.metrics import accuracy_score, precision_score, recall_score

y_pr = md1.predict(xt_sc)
prec = precision_score(yt, y_pr)
rec = recall_score(yt, y_pr)

w = md1.coef_[0]
b = md1.intercept_[0]


p0 = np.mean(y_tr == 0) # число объектов класса 1/ на все 
p1 = np.mean(y_tr == 1)

log_p0 = gss_0.score_samples(xt_sc) + np.log(p0) #  вероятность появления X0 в классе Y0  ln( сумма( w_j * gamma_j ) )  +   ln( P(Класс 0) )
log_p1 = gss_1.score_samples(xt_sc) + np.log(p1)
yg_pr = (log_p1 > log_p0).astype(int) # astype(int) переводит логические ответы в обычные числа: True превращается в число 1 (Класс 1). False превращается в число 0 (Класс 0).

if accuracy_score(yt, yg_pr) < 0.5:
    yg_pr = 1 - yg_pr

prec_g = precision_score(yt, yg_pr)
rec_g = recall_score(yt, yg_pr)

ygc_pr = gpc.predict(xt_sc)

prec_gp = precision_score(yt, ygc_pr)
rec_gp = recall_score(yt, ygc_pr)
probs = gpc.predict_proba(xt_sc)[:, 1]

```

```python
x_min, x_max = x_tr_sc[:, 0].min(), x_tr_sc[:, 0].max()
y_min, y_max = x_tr_sc[:, 1].min(), x_tr_sc[:, 1].max()

xx, yy = np.meshgrid(np.linspace(x_min, x_max, 25), np.linspace(y_min, y_max, 25))
zz = -(w[0] * xx + w[1] * yy + b) / w[2]


fig = go.Figure()


fig.add_trace(go.Scatter3d(
    x=x_tr_sc[y_tr == 0, 0],
    y=x_tr_sc[y_tr == 0, 1],
    z=x_tr_sc[y_tr == 0, 2],
    mode='markers',
    marker=dict(size=4, color='royalblue', opacity=0.7),
    name='Класс 0 (Выборка 1)'
))

fig.add_trace(go.Scatter3d(
    x=x_tr_sc[y_tr == 1, 0],
    y=x_tr_sc[y_tr == 1, 1],
    z=x_tr_sc[y_tr == 1, 2],
    mode='markers',
    marker=dict(size=4, color='darkorange', opacity=0.7),
    name='Класс 1 (Выборки 2 и 3)'
))

fig.add_trace(go.Surface(
    x=xx,
    y=yy,
    z=zz,
    opacity=0.4,
    colorscale=[[0, 'gray'], [1, 'gray']],  # Серый цвет плоскости
    showscale=False,                        # Скрыть цветовую шкалу
    name='Плоскость SVM',
    showlegend=True
))

fig.update_layout(
    title=f'SVM на отмасштабированных данных<br><sup>precision = {prec:.4f}, recall = {rec:.4f}</sup>',
    scene=dict(
        xaxis_title='X (scaled)',
        yaxis_title='Y (scaled)',
        zaxis_title='Z (scaled)'
    ),
    width=900,
    height=700,
    margin=dict(l=0, r=0, b=0, t=50)
)


fig.show()

# Вывод метрик в консоль
print(f'precision = {prec:.6f}, recall = {rec:.6f}')
```

```python
def get_dome_mesh(mean, cov, n_std=1.8, n_pts=30):
    vals, vecs = np.linalg.eigh(cov)
    u = np.linspace(0, 2 * np.pi, n_pts)
    v = np.linspace(0, np.pi, n_pts)
    x = np.outer(np.cos(u), np.sin(v))
    y = np.outer(np.sin(u), np.sin(v))
    z = np.outer(np.ones_like(u), np.cos(v))
    sphere = np.stack([x, y, z], axis=-1)
    
    radii = n_std * np.sqrt(np.maximum(vals, 0))
    dome = (sphere * radii) @ vecs.T + mean
    return dome[:, :, 0], dome[:, :, 1], dome[:, :, 2]


# ==============================================================
# 3. Визуализация в Plotly
# ==============================================================
fig = go.Figure()

# Точки Класса 0
fig.add_trace(go.Scatter3d(
    x=x_tr_sc[y_tr == 0, 0], y=x_tr_sc[y_tr == 0, 1], z=x_tr_sc[y_tr == 0, 2],
    mode='markers', marker=dict(size=4, color='royalblue', opacity=0.7),
    name='Класс 0 (Выборка 1)'
))

# Точки Класса 1
fig.add_trace(go.Scatter3d(
    x=x_tr_sc[y_tr == 1, 0], y=x_tr_sc[y_tr == 1, 1], z=x_tr_sc[y_tr == 1, 2],
    mode='markers', marker=dict(size=4, color='darkorange', opacity=0.7),
    name='Класс 1 (Выборки 2 и 3)'
))

# ----------------- КУПОЛ ДЛЯ КЛАССА 0 (1 штука) -----------------
x0, y0, z0 = get_dome_mesh(gss_0.means_[0], gss_0.covariances_[0])
fig.add_trace(go.Surface(
    x=x0, y=y0, z=z0,
    colorscale=[[0, 'blue'], [1, 'blue']],
    opacity=0.3, showscale=False,
    name='Купол Класса 0', showlegend=True
))

# ----------------- КУПОЛА ДЛЯ КЛАССА 1 (2 штуки) ----------------
# 1-я гауссиана 2-го класса (накроет выборку 2)
x1_a, y1_a, z1_a = get_dome_mesh(gss_1.means_[0], gss_1.covariances_[0])
fig.add_trace(go.Surface(
    x=x1_a, y=y1_a, z=z1_a,
    colorscale=[[0, 'orange'], [1, 'orange']],
    opacity=0.3, showscale=False,
    name='Купол 1 (Класс 1)', showlegend=True
))

# 2-я гауссиана 2-го класса (накроет выборку 3)
x1_b, y1_b, z1_b = get_dome_mesh(gss_1.means_[1], gss_1.covariances_[1])
fig.add_trace(go.Surface(
    x=x1_b, y=y1_b, z=z1_b,
    colorscale=[[0, 'red'], [1, 'red']],
    opacity=0.3, showscale=False,
    name='Купол 2 (Класс 1)', showlegend=True
))

# Оформление и вывод метрик в заголовок
fig.update_layout(
    title=f'GMM: 1 Гауссиана для Класса 0 и 2 Гауссианы для Класса 1<br><sup>precision = {prec_g:.4f}, recall = {rec_g:.4f}</sup>',
    scene=dict(
        xaxis_title='X (scaled)',
        yaxis_title='Y (scaled)',
        zaxis_title='Z (scaled)'
    ),
    width=950,
    height=750,
    margin=dict(l=0, r=0, b=0, t=60)
)
```

```python
def create_rbf_dome(data, n_std=1.8, n_pts=35):
    mean = np.mean(data, axis=0)
    cov = np.cov(data, rowvar=False)
    vals, vecs = np.linalg.eigh(cov)
    
    u = np.linspace(0, 2 * np.pi, n_pts)
    v = np.linspace(0, np.pi, n_pts)
    x = np.outer(np.cos(u), np.sin(v))
    y = np.outer(np.sin(u), np.sin(v))
    z = np.outer(np.ones_like(u), np.cos(v))
    sphere = np.stack([x, y, z], axis=-1)
    
    radii = n_std * np.sqrt(np.maximum(vals, 0))
    dome = (sphere * radii) @ vecs.T + mean
    return dome[:, :, 0], dome[:, :, 1], dome[:, :, 2]


# Выделяем точки для построения экспоненциальных куполов
pts_c0 = x_tr_sc[y_tr == 0]  # Класс 0 (Выборка 1)
pts_c1 = x_tr_sc[y_tr == 1]  # Класс 1 (Выборки 2 и 3)

# Считаем форму куполов
x_d0, y_d0, z_d0 = create_rbf_dome(pts_c0, n_std=1.8)
x_d1, y_d1, z_d1 = create_rbf_dome(pts_c1, n_std=1.8)


# ==============================================================
# 3. 3D-график в Plotly
# ==============================================================
fig = go.Figure()

# --- Точки Класса 0 (предсказанные или истинные) ---
fig.add_trace(go.Scatter3d(
    x=xt_sc[yg_pr == 0, 0],
    y=xt_sc[yg_pr == 0, 1],
    z=xt_sc[yg_pr == 0, 2],
    mode='markers',
    marker=dict(size=4, color='royalblue', opacity=0.8),
    name='Класс 0 (предсказано)'
))

# --- Точки Класса 1 (предсказанные или истинные) ---
fig.add_trace(go.Scatter3d(
    x=xt_sc[yg_pr == 1, 0],
    y=xt_sc[yg_pr == 1, 1],
    z=xt_sc[yg_pr == 1, 2],
    mode='markers',
    marker=dict(size=4, color='darkorange', opacity=0.8),
    name='Класс 1 (предсказано)'
))

# --- Объемная экспонента (купол RBF) для Класса 0 ---
fig.add_trace(go.Surface(
    x=x_d0, y=y_d0, z=z_d0,
    colorscale=[[0, 'blue'], [1, 'blue']],
    opacity=0.3,
    showscale=False,
    name='RBF-купол (Класс 0)',
    showlegend=True
))

# --- Объемная экспонента (купол RBF) для Класса 1 ---
fig.add_trace(go.Surface(
    x=x_d1, y=y_d1, z=z_d1,
    colorscale=[[0, 'darkorange'], [1, 'darkorange']],
    opacity=0.3,
    showscale=False,
    name='RBF-купол (Класс 1)',
    showlegend=True
))

# --- Оформление заголовка с метриками ---
fig.update_layout(
    title=f'Гауссов процесс с RBF-ядром (Классификатор)<br><sup>precision = {prec_gp:.4f}, recall = {rec_gp:.4f}</sup>',
    scene=dict(
        xaxis_title='X (scaled)',
        yaxis_title='Y (scaled)',
        zaxis_title='Z (scaled)'
    ),
    width=950,
    height=750,
    margin=dict(l=0, r=0, b=0, t=60)
)

fig.show()
```
